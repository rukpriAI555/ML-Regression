```python
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset, dtype=int, drop_first=True)
```

```python
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
```

```python
independent
```

```python
dependent = dataset[['charges']]
```

```python
dependent
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30, random_state=0)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
#Model creation in RandomForestRF
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
#Importing GridsearchV for checking the model's performance with all the parameters and cross-validation
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
#regressor=RandomForestRegressor(n_estimators=50, criterion='poisson', random_state=0)RF
```

```python
#Assigning all the parameters to param_grid
```

```python
param_grid = {'criterion':['squared_error','absolute_error', 'poisson'],
'max_features': ['sqrt','log2'],
'n_estimators':[10,100]}
```

```python
#Passing all the parameters, RandomForesgRegressor to gridsearchcv to do gridsearch and cross validation. Refit=True, will get the best model
```

```python
#If Refit=False, GridsearchCv will pull the last model and its parameters
```

```python
grid = GridSearchCV(RandomForestRegressor(), param_grid, refit = True, verbose = 3,n_jobs=-1)
```

```python
#regressor.fit(x_train,y_train)RF
```

```python
#Grid.Fit method below creates the best model
```

```python
grid.fit(x_train, y_train)
```

```python
#cv_results_ will have the result of Gridsearchcv
```

```python
re=grid.cv_results_
```

```python
#y_pred=regressor.predict(x_test)RF
```

```python
#Evaluation Metrics
```

```python
grid_predictions = grid.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r_score=r2_score(y_test,grid_predictions)
```

```python
#r_score
```

```python
print("The R_score value for best parameter {}:".format(grid.best_params_),r_score)
```
