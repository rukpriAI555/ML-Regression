```python
#importing pandas library for accessing database and reading the csv file, also assigning the library to a variable 'pd' for easiness in coding
```

```python
import pandas as pd
```

```python
#Using the read_csv function in pandas/pd library to read the csv file and assigning it to a variable 'dataset'
```

```python
dataset = pd.read_csv("Salary_Data.csv")
```

```python
dataset
```

```python
#by using the variable 'dataset', viewing the data in the csv file above
```

```python
#Assigning independent variable to the input 'YearsExperience'
```

```python
independent = dataset[["YearsExperience"]]
```

```python
independent
```

```python
#viewing the input YearsExperience using the independent variable above
```

```python
dependent= dataset[["Salary"]]
```

```python
#Assigning output 'Salary' to the variable 'dependent' above
```

```python
dependent
```

```python
#Viewing the output 'Salary' using the variable 'dependent' above
```

```python
#Splitting the Training dataset and Test dataset below importing the library sklearn of ML; package model_selection and function train_test_split
```

```python
from sklearn.model_selection import train_test_split
```

```python
#Assigning training input, output, test input and output values to four variables x_train, x_test, y_train and y_test
```

```python
x_train, x_test, y_train, y_test = train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
#Model creation - Learning PhaseI
```

```python
#using sklearn library, linear_model package, linear regression class creating model below
```

```python
from sklearn.linear_model import LinearRegression
```

```python
#Assigning LinearRegression class functions to variable 'regressor'
```

```python
regressor = LinearRegression()
```

```python
#creating the model using fit method of LinearRegression passing training dataset; x_train and y_train
```

```python
regressor.fit(x_train,y_train)
```

```python
#Finding 'weight' and 'bais' and viewing them by calling the associated variables
```

```python
weight = regressor.coef_
```

```python
weight
```

```python
bais=regressor.intercept_
```

```python
bais
```

```python
#Model prediction using test input and predict method of regressor
```

```python
y_pred=regressor.predict(x_test)
```

```python
#Evaluation Metrics to evaluate the correctness of the Model using RSquare function from metrics package and sklearn library
```

```python
from sklearn.metrics import r2_score
```

```python
#Assigning RSquare function to variable r_score, passing actual output and predicted output as parameters to r2_score function
```

```python
r_score=r2_score(y_test, y_pred)
```

```python
#Viewing the r_score
```

```python
r_score
```

```python
#Since the r_score is close to 1, the model is good and can be saved
```

```python
#Importing pickle package for saving the model
```

```python
import pickle
```

```python
#Assigning a file name for the model
```

```python
filename = "finalized_model_linear.sav"
```

```python
#saving the model against the file; using dump function and passing regressor;writing the model in the file created
```

```python
pickle.dump(regressor, open(filename, 'wb'))
```
